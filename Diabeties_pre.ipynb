```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
df[["Glucose","BMI","Insulin","Age"]]=scaler.fit_transform(df[["Glucose","BMI","Insulin","Age"]])
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.iloc[:,0:-1],
                                                df.iloc[:,-1],
                                                test_size=0.2,
                                                random_state=0)
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(drop="first",sparse_output=False,dtype=np.int32)

newX_train=ohe.fit_transform(X_train[['Gender',"Diet_Type","Exercise_Frequency","Heredity","Smoking","Alcohol"]])
newX_test=ohe.transform(X_test[['Gender',"Diet_Type","Exercise_Frequency","Heredity","Smoking","Alcohol"]])
```

```python
newX_test
```

```python
newX_train,newX_train.shape
```

```python
new_dfXt=np.hstack((X_train.drop(['Gender',"Diet_Type","Exercise_Frequency","Heredity","Smoking","Alcohol"],axis=1).values, newX_train))
```

```python
pd.DataFrame(new_dfXt)
```

```python
new_dfXte=np.hstack((X_test.drop(['Gender',"Diet_Type","Exercise_Frequency","Heredity","Smoking","Alcohol"],axis=1).values, newX_test))
```

```python
pd.DataFrame(new_dfXte)
```

```python
y_train
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
lr.fit(new_dfXt,y_train)
```

```python
y_pred=lr.predict(new_dfXte)
```

```python
from sklearn.metrics import accuracy_score
```

```python
Score=accuracy_score(y_test,y_pred)
print(f"Accuracy Score is:{Score}")
```
